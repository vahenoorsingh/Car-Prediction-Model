```python
car_df['match_city'] = car_df['Location'].str.lower()
indian_cood['match_city'] = indian_cood['city'].str.strip().str.lower()

lat_map = indian_cood.set_index('match_city')['lat'].to_dict()
lng_map = indian_cood.set_index('match_city')['lng'].to_dict()

car_df['Latitude'] = car_df['match_city'].map(lat_map)
car_df['Longitude'] = car_df['match_city'].map(lng_map)


car_df.head()

for city, (lat, lng) in missing_coords.items():
    lat_map[city] = lat
    lng_map[city] = lng


car_df['Latitude'] = car_df['match_city'].map(lat_map)
car_df['Longitude'] = car_df['match_city'].map(lng_map)

car_df = car_df.drop(columns=['match_city'])

car_df.head()
```

```python
city_df = car_df['Location'].value_counts().reset_index()
city_df.columns = ['Location', 'Car_Count']
city_df.head()

city_df['match_city'] = city_df['Location'].str.lower()
indian_cood['match_city'] = indian_cood['city'].str.strip().str.lower()

lat_map = indian_cood.set_index('match_city')['lat'].to_dict()
lng_map = indian_cood.set_index('match_city')['lng'].to_dict()

city_df['Latitude'] = city_df['match_city'].map(lat_map)
city_df['Longitude'] = city_df['match_city'].map(lng_map)


city_df.head()

for city, (lat, lng) in missing_coords.items():
    lat_map[city] = lat
    lng_map[city] = lng


city_df['Latitude'] = city_df['match_city'].map(lat_map)
city_df['Longitude'] = city_df['match_city'].map(lng_map)

city_df = city_df.drop(columns=['match_city'])

avg_price_df = car_df.groupby('Location')['Price'].mean().reset_index()
avg_price_df.columns = ['Location', 'Average_Price']

city_df = city_df.merge(avg_price_df, on='Location', how='left')
city_df['Average_Price'] = city_df['Average_Price'].round(2)
print(city_df.head(10))
city_df.plot(kind="scatter", x="Longitude", y="Latitude", grid=True, figsize=(6,6), c="Average_Price", cmap="jet")
#i dont think it was worth the effort 😭
```

Mapping owners

```python
owner_mapping = {
    'UnRegistered Car': 0,
    'First': 1,
    'Second': 2,
    'Third': 3,
    'Fourth': 4,
    '4 or More': 5
}

car_df['Owner'] = car_df['Owner'].map(owner_mapping)

car_df.head()
```

```python
car_df['Color'].unique().value_counts()

from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False, drop='first')

color_encoded = encoder.fit_transform(car_df[['Color']])

encoded_cols = encoder.get_feature_names_out(['Color'])
df_encoded = pd.DataFrame(color_encoded, columns=encoded_cols, index=car_df.index)

car_df = pd.concat([car_df, df_encoded], axis=1).drop(columns=['Color'])

car_df.head()
```
